```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data=pd.read_csv("customers_data.csv")
df=pd.DataFrame(data)

df["Age"].fillna(df["Age"].median(),inplace=True)
df["Income"].fillna(df["Income"].mean(),inplace=True)

df["Gender"].fillna(df["Gender"].mode()[0],inplace=True)
df["City"].fillna(df["City"].mode()[0],inplace=True)
df["Date_of_Joining"].fillna("01-01-2020",inplace=True)
        
df.drop_duplicates(inplace=True)

df["Date_of_Joining"]=pd.to_datetime(df["Date_of_Joining"])
df["Gender"]=df["Gender"].astype("category")

Q1=df["Income"].quantile(0.25)
Q3=df["Income"].quantile(0.75)
IQR=Q3-Q1
df[(df["Income"]>=Q1-1.5*IQR) & (df["Income"]<=Q3+1.5*IQR)]
df=pd.get_dummies(df,columns=["City"],drop_first=True)

from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])

from sklearn.preprocessing import StandardScaler
st=StandardScaler()
df[["Age"]]=st.fit_transform(df[["Age"]])
df[["Income"]]=st.fit_transform(df[["Income"]])
```
